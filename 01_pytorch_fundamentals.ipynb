```python
#torch.squeeze() -remove all single dimension from large tensor
print(f'Previous tensor:{x_reshaped}')
print(f'Previous shape:{x_reshaped.shape}')
x_reshaped=x.squeeze()
print(f'New tensor:{x_reshaped}')
print(f'New shape:{x_reshaped.shape}')
```

```python
#torch.unsqueeze()-adds a single dimension to  a target tensor at a specific dimension
print(f'Previous tensor:{x_reshaped}')
print(f'Previous shape:{x_reshaped.shape}')
x_unsqueezed=x_reshaped.unsqueeze(dim=1)
print(f'New tensor:{x_reshaped}')
print(f'New shape:{x_reshaped.shape}')
```

```python
#torch.unsqueeze()-adds a single dimension to  a target tensor at a specific dimension
print(f'Previous tensor:{x_unsqueezed}')
print(f'Previous shape:{x_unsqueezed.shape}')
x_unsqueezed=x_unsqueezed.unsqueeze(dim=1)
print(f'New tensor:{x_unsqueezed}')
print(f'New shape:{x_unsqueezed.shape}')
```

```python
#torch.permute - rearranges the dimension of a target tensor in a specified order
x=torch.randn(2,3,5)
x.size()
torch.permute(x,(2,0,1)).size()
```

```python
#torch.permute - rearranges the dimension of a target tensor in a specified order

x_original=torch.rand(224,224,3) #height widht color channels
#permute the original tensor

x_permute=x_original.permute(2,0,1) #shifts acis 0->1 1->2 2->0
x_original.size()

print(f"original tensor size {x_original.size()}")
print(f"Changed tensor size {x_permute.size()}")
```

###Pytorch tensor and numpy

Numpy is a popular scientific numerical librarry


And because of this pytorch has functionality to interact with it

* Data in Numpy, want in pytorch tensor rom numpty `torch.from_numpy(ndarray)`

* Pytorch tensor->numpy `torch.Tensor.numpy()`

#warining when converting tensor from numpy default dtype was float64

```python
#Pytorch and numpy
import torch
import numpy as np
```

```python
array=np.arange(1.0,10.0)
```

```python
tensor=torch.from_numpy(array)
array,tensor
```

```python
array.dtype
```

```python
tensor.dtype
```

```python
tensor=torch.from_numpy(array).type(torch.float32)
tensor
```

```python
array+1,tensor
```

```python
#Tensor to numpy
tensor=torch.ones(7)
numpy_tensor=tensor.numpy()
tensor,numpy_tensor
```

```python
tensor=tensor+1
tensor,numpy_tensor
```
