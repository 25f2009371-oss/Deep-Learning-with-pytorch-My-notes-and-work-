### Random tensor


Random tensors are important because the way many neural network learn is that they start with tensors full of random numbers and adjusts those random numbers to better represent the data

`start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers `



```python
#Random tensor

random_tensor=torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor=torch.rand(10,10,10)
random_tensor.ndim
```

###Tensor datatypes



**Note:** Tensor datatype is one of the 3 big errors you will run into with PyTorch & deep learning:

1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device


A gradient represents a gradual transition or rate of change across a distance or mathematical space. The term is heavily relied upon across three major fields: design and web development, mathematics and physics, and geography and engineering.

```python


float_32_tensor=torch.tensor([1.0,3.0,9],
                             dtype=None, #what datatype is the tensors float 32 or float 16 larger the number more space more precision
                             device="cuda", # what device is your tensor on
                             requires_grad=False #whether or not to track gradients with tensor operations
                             )
#float 64 till exist
float_32_tensor
```

```python
#converting 32 float tensor to 16 bit tensor
float_16_tensor=float_32_tensor.type(torch.float16)
float_16_tensor
```
